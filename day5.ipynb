# 🌳 Decision Tree & 🌲 Random Forest

# Decision Tree = A Machine Learning model that makes predictions by asking a sequence of questions and following branches to reach a final answer.

          Credit History?
           /           \
         Good          Bad
          /              \
     Income > 5000?      Reject
       /      \
     Yes       No
     /          \
 Approve       Reject


 Isliye Decision Tree ko tum machine learning ka smart IF-ELSE system samajh sakte ho.

| Term              | Meaning                              |
| ----------------- | ------------------------------------ |
| **Root Node**     | The first question                   |
| **Decision Node** | A question/condition inside the tree |
| **Branch**        | The path taken after a condition     |
| **Leaf Node**     | The final prediction                 |
| **Depth**         | How many levels the tree has         |
